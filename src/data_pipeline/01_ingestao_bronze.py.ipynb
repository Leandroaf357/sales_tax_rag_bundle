```sql
%python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

CSV_PATH = "/Volumes/sales_tax_catalog/bronze/landing_zone/Sales_Tax_Allocation__City.csv"
BRONZE_TABLE = "sales_tax_catalog.bronze.sales_tax_raw"

def ingest_bronze():
    print(f"Lendo dados do arquivo: {CSV_PATH}")
    df = spark.read.csv(CSV_PATH, header=True, inferSchema=True)
    
    if "_rescued_data" in df.columns:
        df = df.drop("_rescued_data")
    
    df.write.format("delta") \
      .mode("overwrite") \
      .option("overwriteSchema", "true") \
      .option("delta.columnMapping.mode", "name") \
      .option("delta.minReaderVersion", "2") \
      .option("delta.minWriterVersion", "5") \
      .saveAsTable(BRONZE_TABLE)
      
    print(f"Dados brutos ingeridos em {BRONZE_TABLE}")

if __name__ == "__main__":
    ingest_bronze()
```
